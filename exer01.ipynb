01)Faça uma função que dado um vetor de inteiros retorne o maior inteiro do vetor

```python
def maior_inteiro(vetor):
    if len(vetor) == 0:
        return None # retorna None se o vetor estiver vazio 
    maior = vetor[0]
    for numero in vetor:
        if numero > maior:
            maior = numero
    return maior 


vetor = [3, 5, 7, 2, 8, 1]
print(maior_inteiro(vetor))
```
